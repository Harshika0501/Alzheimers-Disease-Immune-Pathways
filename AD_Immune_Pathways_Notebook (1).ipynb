## 1. Objective

1. To compile the set of genome-wide significant AD risk genes reported by a large, recent AD GWAS meta-analysis.
2. To independently test, using standard gene-set enrichment tools, whether this gene set shows statistically significant enrichment for immune-system-related biological processes.
3. To evaluate whether an unbiased, independently-run analysis (using different tools and a different statistical approach than the original authors) converges on the same broad conclusion.

## 2. Data source

The gene list (n = 76 unique genes) was compiled from Tables 1 and 2 of:

> Bellenguez, C., Küçükali, F., Jansen, I.E. et al. (2022). New insights into the genetic etiology of Alzheimer's disease and related dementias. *Nature Genetics*, 54, 412–436. https://doi.org/10.1038/s41588-022-01024-z

This study performed a two-stage GWAS meta-analysis of 111,326 clinically diagnosed or proxy AD/related-dementia cases and 677,663 controls, identifying 75 genome-wide significant risk loci (33 previously known, 42 new at the time of publication). The study's own pathway enrichment analysis found that, beyond the expected amyloid and tau pathway signals, gene sets related to lipid metabolism, endocytosis, and immunity (including macrophage and microglial cell activation) were also significantly enriched.

One entry, "IGH gene cluster," was excluded from the working gene list, due to its genomic complexity as a result of known immunoglobulin gene fusion events.

## 3. Methods

### 3.1 Gene set compilation
All genes listed in Table 1 (known loci) and Table 2 (new loci at time of publication) were extracted. Genes with multiple independent genome-wide significant SNPs reported in the source tables (TREM2 ×3, PLCG2 ×2, SORL1 ×2, SLC24A4 ×2, MME ×2) were deduplicated to a single gene entry each, yielding **76 unique gene symbols**.

### 3.2 Pathway enrichment analysis
The 76-gene list was submitted to **Enrichr** (Chen et al., 2013; Kuleshov et al., 2016), an independent, web-based gene set enrichment tool, and tested against three gene-set libraries:
- **GO Biological Process**
- **KEGG**
- **Reactome**

For each library, Enrichr computes enrichment using a Fisher's exact test against a genome-wide background, and reports both a raw p-value and a Benjamini–Hochberg-corrected adjusted p-value (to control the false discovery rate across the many simultaneous term comparisons performed). Terms with adjusted p-value < 0.05 were retained as statistically significant.

Significant terms were then screened for immune relevance using a fixed keyword list — immune, inflammation, microglia, macrophage, phagocytosis, complement, cytokine, leukocyte, TNF — chosen in advance based on the terminology used in the source paper's own pathway discussion, to reduce post-hoc selection bias. Genes underlying immune-flagged terms were pooled across all three libraries (union) and deduplicated, then manually reviewed to confirm that their primary annotated function was immune-related rather than an incidental match to a narrow GO term.

### 3.3 Protein–protein interaction network analysis
The same full 76-gene list (not the immune-flagged subset) was submitted to **STRING v12** (Szklarczyk et al., 2023) to test, independently of the enrichment analysis above, whether immune-relevant genes form a distinguishable, densely interconnected network module based on documented protein–protein interaction evidence (co-expression, curated databases, experimental data, and text-mining).
STRING's built-in functional enrichment output (its own implementation of GO term enrichment, computed via a different backend than Enrichr) was used as a secondary, independent statistical cross-check.

### 3.4 Cross-validation against the source paper
The final immune-flagged gene set was compared against the specific genes the Bellenguez et al. (2022) paper explicitly discussed as immune/microglia-related in its main text (TREM2, RHOH, BLNK, SIGLEC11, LILRB2, SHARPIN, RBCK1, OTULIN, ADAM17, TNIP1, SPPL2A), to assess convergence between this independent analysis and the original authors' own conclusions.

### 4.3 STRING network analysis

## 5. Discussion

### 5.1 Why immune involvement in AD genetics is biologically plausible

The genes identified as immune-relevant in this analysis map onto several well-characterized mechanisms in the neuroimmunology literature, lending biological interpretability to the statistical result:

**Microglial receptor signaling and the disease-associated microglia (DAM) phenotype.** TREM2 is a microglial surface receptor central to this analysis. In response to amyloid-β and cellular debris, TREM2 signaling (via the DAP12 adaptor complex) drives microglia to transition into a distinct transcriptional and functional state — "disease-associated microglia" — associated with enhanced phagocytic clearance of amyloid (Deczkowska et al., 2018; see also review by Li & Zhang, 2018, and a recent review of TREM2 mechanisms and therapeutic implications). TREM2's genetic risk signal is among the strongest non-APOE effects identified in AD GWAS, consistent with its position as the largest, most gene-dense node in the STRING immune cluster in this analysis.

**Complement-mediated synaptic pruning.** CR1 (complement receptor 1) and CLU (clusterin, a complement regulatory protein) both appeared in the immune-flagged set. The classical complement cascade (C1q, C3, and their receptors) tags synapses for microglial engulfment during normal brain development; foundational work by Stevens, Hong, and colleagues showed this same mechanism is reactivated in AD, where excessive complement-mediated pruning contributes to the synapse loss that correlates most strongly with cognitive decline (Hong et al., 2016, *Science*; Stevens et al., 2007, *Cell*). This gives CR1's presence in the gene list a specific, mechanistic interpretation beyond a generic "immune" label.

**Antigen presentation.** HLA-DQA1, part of the major histocompatibility complex (MHC) class II machinery, was independently flagged in this analysis and was also the subject of a dedicated supplementary analysis in Bellenguez et al. This gene's presence reflects the adaptive, antigen-presentation arm of the immune signal, distinct from the innate/microglial genes discussed above.

**TNF-α signaling and the LUBAC complex.** ADAM17, TNIP1, SPPL2A, SHARPIN, and RBCK1 all converge on tumor necrosis factor-alpha (TNF-α) signaling — a pathway Bellenguez et al. specifically highlighted via their identification of the linear ubiquitin chain assembly complex (LUBAC, formed by SHARPIN, RBCK1, and regulated by OTULIN) as a novel genetic finding in AD. TNF-α signaling has documented roles in both neuroinflammation and normal synaptic physiology, illustrating why immune signaling genes can have dual, context-dependent roles in the brain.

**Progranulin (GRN) as a case study in the amyloid–immune interface.** GRN was flagged in this analysis and, notably, is explicitly discussed by Bellenguez et al. as a TNF receptor ligand and antagonist of TNF-α signaling — despite not being one of the 11 genes named in their summary "immune gene" list. This illustrates a broader point: the paper's own supporting evidence for immune involvement is broader than the specific gene names it chose to foreground in its main narrative.

### 5.2 Relevance to vaccinology and immunotherapy

This neuroimmune signal is not merely descriptive — it underlies a substantial and clinically active area of immunotherapeutic development in AD, which is directly relevant to a transition toward vaccinology. Active and passive amyloid-β immunization has been attempted since the first trial of an amyloid vaccine, AN1792, in 2000–2001; this trial was halted after a subset of patients developed T-cell-mediated meningoencephalitis, but long-term follow-up showed that antibody responders had genuine amyloid plaque clearance and, in some analyses, reduced functional decline (Gilman et al., 2005; Nicoll et al., 2019). This history directly informed the design of later, safer second-generation active vaccines (e.g., CAD106) and the current generation of approved passive monoclonal antibody immunotherapies (e.g., lecanemab, donanemab), which work by directing the adaptive immune system against amyloid — a strategy whose success and side-effect profile (notably ARIA, amyloid-related imaging abnormalities, itself thought to involve vascular immune/inflammatory responses) is inseparable from the innate immune/microglial biology this project's gene list independently recovered.


### 5.5 Limitations

- Keyword-based term screening is not a formal traversal of the GO ontology's hierarchical structure; genes could be missed if their associated term names lack an exact keyword match, and this analysis does not use tools such as QuickGO/AmiGO that would allow a structurally rigorous check against the formal GO:0002376 ("immune system process") subtree.
- Enrichment analysis identifies statistical over-representation of annotated gene function; it does not constitute experimental validation that these genes causally act through immune mechanisms specifically in the context of AD pathology.
